```python
class TrafficDevice:
    def activate(self):
        print("Traffic device activated")

class TrafficLight(TrafficDevice):
    def deactivate(self):
        print(f"Traffic device activated")


class SpeedCamera(TrafficDevice):
    def activate(self):
        print(f"Speed Camera activated")

class PedestrianSignal(TrafficDevice):
    def activate(self):
        print(f"Pedestrian Signal activated")

class EmergencySiren(TrafficDevice):
    def activate(self):
        print(f"Emergency Siren activated")

devices = [
    TrafficLight(),
    SpeedCamera(),
    PedestrianSignal(),
    EmergencySiren(),
]
for device in devices:
    device.activate()








```
